```python
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)

pd.DataFrame({'Jumlah Missing': missing, 'Persentase (%)': missing_pct}).sort_values('Jumlah Missing', ascending=False)
```

```python
kolom_numerik_missing = df.select_dtypes(include=[np.number]).columns[df.select_dtypes(include=[np.number]).isnull().any()]

for kol in kolom_numerik_missing:
    median_val = df[kol].median()
    df[kol] = df[kol].fillna(median_val)
    print(f"Kolom '{kol}' diisi dengan median = {median_val}")

print("\nJumlah missing value setelah imputasi:")
print(df.isnull().sum().sum())
```

```python
jumlah_duplikat = df.duplicated().sum()
print(f"Jumlah baris duplikat: {jumlah_duplikat}")

df = df.drop_duplicates()
print(f"Jumlah baris setelah penghapusan duplikat: {df.shape[0]}")
```

```python
plt.figure(figsize=(8, 2))
sns.boxplot(x=df['student_id'], color='salmon')
plt.title('Boxplot student_id - Deteksi Outlier')
plt.show()

Q1 = df['student_id'].quantile(0.25)
Q3 = df['student_id'].quantile(0.75)
IQR = Q3 - Q1
batas_bawah = Q1 - 1.5 * IQR
batas_atas = Q3 + 1.5 * IQR

outlier = df[(df['student_id'] < batas_bawah) | (df['student_id'] > batas_atas)]
print(f"Batas bawah wajar: {batas_bawah:.0f}")
print(f"Batas atas wajar : {batas_atas:.0f}")
print(f"Jumlah data outlier ditemukan: {outlier.shape[0]}")
outlier[['student_id']]
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['gpa'], bins=20, kde=True, color='steelblue')
plt.title('Distribusi IPK (GPA) Mahasiswa')
plt.xlabel('GPA')
plt.ylabel('Frekuensi')
plt.show()

kolom_numerik = ['age', 'attendance_pct', 'assignment_score', 'midterm_score',
                 'final_score', 'study_hours_week', 'lms_activity_pct',
                 'sleep_hours', 'gpa', 'internet_quota_gb']
df[kolom_numerik].hist(figsize=(15, 10), bins=15, color='cornflowerblue', edgecolor='black')
plt.suptitle('Distribusi Seluruh Variabel Numerik', fontsize=14)
plt.tight_layout()
plt.show()

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='status', hue='status', palette='Set2', legend=False)
plt.title('Distribusi Status Kelulusan Mahasiswa')
plt.xlabel('Status')
plt.ylabel('Jumlah Mahasiswa')
plt.show()

print(df['status'].value_counts(normalize=True).round(3) * 100, "%")

plt.figure(figsize=(8, 4))
sns.countplot(data=df, y='study_program', hue='study_program',
              order=df['study_program'].value_counts().index, palette='viridis', legend=False)
plt.title('Distribusi Mahasiswa per Program Studi')
plt.xlabel('Jumlah Mahasiswa')
plt.ylabel('Program Studi')
plt.show()
```

```python
plt.figure(figsize=(6, 5))
sns.boxplot(data=df, x='status', y='gpa', hue='status', palette='Set2', legend=False)
plt.title('Perbandingan GPA berdasarkan Status Kelulusan')
plt.xlabel('Status')
plt.ylabel('GPA')
plt.show()

plt.figure(figsize=(7, 5))
sns.scatterplot(data=df, x='attendance_pct', y='gpa', hue='status', alpha=0.6, palette='Set1')
plt.title('Hubungan Persentase Kehadiran dengan GPA')
plt.xlabel('Persentase Kehadiran (%)')
plt.ylabel('GPA')
plt.show()

rata_gpa_prodi = df.groupby('study_program')['gpa'].mean().sort_values(ascending=False)
print(rata_gpa_prodi)

plt.figure(figsize=(8, 4))
rata_gpa_prodi.plot(kind='bar', color='teal')
plt.title('Rata-rata GPA per Program Studi')
plt.ylabel('Rata-rata GPA')
plt.xticks(rotation=30, ha='right')
plt.show()

plt.figure(figsize=(7, 5))
sns.regplot(data=df, x='study_hours_week', y='final_score',
                scatter_kws={'alpha': 0.4}, line_kws={'color' : 'red'})
plt.title('Hubungan jam belajar mandiri per minggu dengan nilai akhir')
plt.xlabel('Jam Belajar per Minggu')
plt.ylabel('Nilai Akhir (UAS)')
plt.show()
```

```python
kolom_korelasi = ['age', 'attendance_pct', 'assignment_score', 'midterm_score',
                  'final_score', 'study_hours_week', 'lms_activity_pct',
                  'sleep_hours', 'gpa', 'internet_quota_gb']

korelasi = df[kolom_korelasi].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(korelasi, annot=True, fmt='.2f', cmap='coolwarm', square=True, linewidths=0.5)
plt.title('Heatmap Korelasi Antar Variabel Numerik')
plt.show()

subset_kolom = ['attendance_pct', 'study_hours_week', 'gpa', 'final_score', 'status']

sns.pairplot(df[subset_kolom], hue='status', palette='Set1', diag_kind='kde', corner=True)
plt.suptitle('Pairplot Antar Variabel Kunci berdasarkan Status Kelulusan', y=1.02)
plt.show()

ringkasan = df.groupby(['status', 'gender'])[['gpa', 'attendance_pct', 'study_hours_week']].mean().round(2)
ringkasan
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='lms_activity_pct', y='assignment_score', hue='status', alpha=0.7, palette='Set1')
sns.regplot(data=df, x='lms_activity_pct', y='assignment_score', scatter=False, color='red', line_kws={'linestyle':'--'})
plt.title('Hubungan Keaktifan LMS dengan Nilai Tugas')
plt.xlabel('Keaktifan LMS (%)')
plt.ylabel('Nilai Tugas')
plt.show()

korelasi_lms_tugas = df['lms_activity_pct'].corr(df['assignment_score'])
print(f"Korelasi LMS vs Nilai Tugas: {korelasi_lms_tugas:.2f}")
```

```python
plt.figure(figsize=(8, 2.5))
sns.boxplot(x=df['internet_quota_gb'], color='lightgreen')
plt.title('Boxplot Kuota Internet (GB) - Deteksi Outlier')
plt.xlabel('Kuota Internet (GB)')
plt.show()

Q1 = df['internet_quota_gb'].quantile(0.25)
Q3 = df['internet_quota_gb'].quantile(0.75)
IQR = Q3 - Q1
batas_bawah = Q1 - 1.5 * IQR
batas_atas = Q3 + 1.5 * IQR

outlier_quota = df[(df['internet_quota_gb'] < batas_bawah) | (df['internet_quota_gb'] > batas_atas)]

print("--- Hasil Perhitungan IQR ---")
print(f"Batas Bawah Wajar : {batas_bawah:.2f} GB")
print(f"Batas Atas Wajar  : {batas_atas:.2f} GB")
print(f"Jumlah Outlier    : {outlier_quota.shape[0]} data")
```
